```python

from langchain_groq import ChatGroq
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough

# 1. Create the updated language model
llm = ChatGroq(
    api_key=GROQ_API_KEY,
    model="openai/gpt-oss-120b"
)

# 2. Create the PDF retriever
retriever = vectorstore.as_retriever(
    search_kwargs={"k": 3}
)

# 3. Combine the retrieved PDF chunks
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)

# 4. Create the RAG prompt
rag_prompt = ChatPromptTemplate.from_template(
    """
    Answer the question using only the context below.
    If the answer is not available in the context,
    say "I don't know."

    Context:
    {context}

    Question:
    {question}

    Answer:
    """
)

# 5. Create the RAG chain
rag_chain = (
    {
        "context": retriever | format_docs,
        "question": RunnablePassthrough()
    }
    | rag_prompt
    | llm
    | StrOutputParser()
)

print("PDF RAG chain created successfully!")

# 6. Ask a question about your PDF
question = "What is the main topic of this PDF?"

answer = rag_chain.invoke(question)

print("\nQuestion:", question)
print("\nAnswer:")
print(answer)
```

```python

import gradio as gr
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_chroma import Chroma

# Function to load a PDF and create its RAG chain
def load_pdf(pdf_file):
    global rag_chain

    if pdf_file is None:
        return "⚠️ Please upload a PDF first!"

    try:
        # Load the uploaded PDF
        loader = PyPDFLoader(pdf_file)
        documents = loader.load()

        if not any(doc.page_content.strip() for doc in documents):
            return "⚠️ No readable text found. Try a text-based PDF."

        # Split the PDF into chunks
        splitter = RecursiveCharacterTextSplitter(
            chunk_size=300,
            chunk_overlap=50
        )
        chunks = splitter.split_documents(documents)

        # Create embeddings using your existing model
        vectorstore = Chroma.from_documents(
            documents=chunks,
            embedding=embeddings,
            collection_name="uploaded_pdf_rag"
        )

        # Create a retriever
        retriever = vectorstore.as_retriever(
            search_kwargs={"k": 3}
        )

        # Create the RAG chain
        rag_chain = (
            {
                "context": retriever | format_docs,
                "question": lambda x: x
            }
            | rag_prompt
            | llm
            | StrOutputParser()
        )

        return (
            f"✅ PDF loaded successfully!\n"
            f"📄 Pages: {len(documents)}\n"
            f"🧩 Text chunks: {len(chunks)}\n"
            "You can now ask questions below."
        )

    except Exception as e:
        return f"❌ Error loading PDF: {str(e)}"


# Function to answer questions
def answer_question(question):
    if "rag_chain" not in globals():
        return "⚠️ Please upload a PDF first."

    if not question or not question.strip():
        return "⚠️ Please enter a question."

    try:
        return rag_chain.invoke(question)
    except Exception as e:
        return f"❌ Error generating answer: {str(e)}"


# Colorful Gradio interface
with gr.Blocks(
    theme=gr.themes.Soft(
        primary_hue="violet",
        secondary_hue="pink",
        neutral_hue="slate"
    ),
    title="PDF RAG Assistant"
) as app:

    gr.Markdown("""
    # 🌈 📚 PDF RAG Assistant
    ### Upload your PDF • Ask questions • Get AI-powered answers

    <div style="background:linear-gradient(90deg,#7c3aed,#ec4899,#f59e0b);
    padding:4px;border-radius:12px;"></div>
    """)

    with gr.Row():
        with gr.Column():
            pdf_input = gr.File(
                label="📂 Upload your PDF",
                file_types=[".pdf"],
                type="filepath"
            )

            upload_button = gr.Button(
                "🚀 Process PDF",
                variant="primary"
            )

            status_output = gr.Textbox(
                label="📋 PDF Status",
                lines=4,
                interactive=False
            )

        with gr.Column():
            question_input = gr.Textbox(
                label="💬 Ask a question",
                placeholder="Example: What is the main topic of this PDF?",
                lines=4
            )

            ask_button = gr.Button(
                "✨ Generate Answer",
                variant="primary"
            )

            answer_output = gr.Textbox(
                label="🤖 AI Answer",
                lines=10,
                interactive=False
            )

    upload_button.click(
        fn=load_pdf,
        inputs=pdf_input,
        outputs=status_output
    )

    ask_button.click(
        fn=answer_question,
        inputs=question_input,
        outputs=answer_output
    )

    gr.Examples(
        examples=[
            ["What is the main topic of this PDF?"],
            ["Explain the important concepts."],
            ["What are the key points?"],
            ["Give a short summary of the document."]
        ],
        inputs=question_input
    )

    gr.Markdown("""
    ---
    💜 **Powered by:** LangChain • Chroma • Hugging Face • Groq

    🎯 Upload a PDF and ask questions based on its content.
    """)

app.launch(share=True)
```
